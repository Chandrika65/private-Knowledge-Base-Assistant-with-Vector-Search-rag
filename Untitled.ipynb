```python
import pandas as pd

from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS
```

```python
import pandas as pd
df=pd.read_csv("customer_support_tickets.csv")
df
```

```python
df.columns
```

```python
df = df.fillna("")

df["text"] = (
    "Ticket Type: " + df["Ticket Type"].astype(str) +
    "\nSubject: " + df["Ticket Subject"].astype(str) +
    "\nDescription: " + df["Ticket Description"].astype(str) +
    "\nResolution: " + df["Resolution"].astype(str) +
    "\nPriority: " + df["Ticket Priority"].astype(str)
)
```

```python
df
```

```python
documents = [
    Document(page_content=row["text"])
    for _, row in df.iterrows()
]
```

```python
documents
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

docs = splitter.split_documents(documents)

print(len(docs))
```

```python
embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vectorstore = FAISS.from_documents(
    docs,
    embedding_model
)
```

```python
vectorstore.save_local('faiss_index')
```
